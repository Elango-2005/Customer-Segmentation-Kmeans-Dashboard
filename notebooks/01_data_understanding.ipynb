## Column Understanding Summary

The **Customers** dataset contains customer demographic and location information. It provides the customer identifiers (`customer_id` and `customer_unique_id`) required for linking customer records with their corresponding orders and performing customer-level analysis.

The **Orders** dataset records the complete lifecycle of each order, including its status and important timestamps. Among these, the `order_purchase_timestamp` is the most significant for this project, as it will be used to calculate the **Recency** feature during feature engineering.

The **Order Items** dataset stores product-level transaction details for each order. The `price` column represents the product cost and will be used to calculate the total amount spent by each customer. Other columns provide additional order and seller information that can be used for extended analysis.

The **Payments** dataset contains payment-related information for each order. The `payment_value` column represents the total amount paid and will be used to compute the **Monetary** value of each customer, which is a key feature for customer segmentation.

Overall, these datasets are connected through the common identifiers `customer_id` and `order_id`, enabling the construction of a unified customer-level dataset. This merged dataset will later be used for feature engineering (Recency, Frequency, and Monetary values) and K-Means clustering to identify distinct customer segments.

Identify Keys & Relationships

## Primary and Foreign Keys

| Dataset | Primary Key | Foreign Key | Purpose |
|----------|-------------|-------------|---------|
| Customers | `customer_id` | — | Stores customer information |
| Orders | `order_id` | `customer_id` | Stores order details for each customer |
| Order Items | (`order_id`, `order_item_id`) | `order_id` | Stores product details for each order |
| Payments | (`order_id`, `payment_sequential`) | `order_id` | Stores payment details for each order |

## Dataset Relationships

The Olist E-Commerce dataset follows a relational database structure where each dataset is connected through common identifiers.

- The **Customers** dataset is linked to the **Orders** dataset using the `customer_id` column.
- Each customer can place multiple orders over time.
- The **Orders** dataset is linked to the **Order Items** dataset using the `order_id` column.
- A single order can contain one or more products.
- The **Orders** dataset is also linked to the **Payments** dataset using the `order_id` column.
- A single order may have one or multiple payment records depending on the payment method.

These relationships allow the construction of a unified customer-level dataset for feature engineering and customer segmentation.

                 Customers
             -----------------
             customer_id (PK)
                     │
                     │
          One Customer → Many Orders
                     │
                     ▼
                 Orders
             -----------------
             order_id (PK)
             customer_id (FK)
                     │
          ┌──────────┴──────────┐
          │                     │
          ▼                     ▼
    Order Items             Payments
------------------      -------------------
order_id (FK)          order_id (FK)
order_item_id          payment_sequential
price                  payment_value

## Relationship Cardinality

| Parent Table | Child Table | Relationship |
|--------------|-------------|--------------|
| Customers | Orders | One-to-Many (1:N) |
| Orders | Order Items | One-to-Many (1:N) |
| Orders | Payments | One-to-Many (1:N) |
