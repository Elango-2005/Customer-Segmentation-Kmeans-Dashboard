Import Libraries

```python
import pandas as pd
import numpy as np
```

Load Cleaned Datasets

```python
customers = pd.read_csv("../datasets/cleaned/customers_clean.csv")

orders = pd.read_csv(
    "../datasets/cleaned/orders_clean.csv",
    parse_dates=[
        "order_purchase_timestamp",
        "order_estimated_delivery_date"
    ]
)

order_items = pd.read_csv("../datasets/cleaned/order_items_clean.csv")

payments = pd.read_csv("../datasets/cleaned/payments_clean.csv")
```

Merge Customers and Orders

Customers (1)
      │
customer_id
      │
Orders (N)

```python
customer_orders = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="inner"
)

print(customer_orders.shape)
customer_orders.head()
```

Merge Order Items

Orders (1)
      │
order_id
      │
Order Items (N)

```python
customer_orders_items = pd.merge(
    customer_orders,
    order_items,
    on="order_id",
    how="inner"
)

print(customer_orders_items.shape)
customer_orders_items.head()
```

Merge Payments

Orders (1)
      │
order_id
      │
Payments (N)

```python
merged_df = pd.merge(
    customer_orders_items,
    payments,
    on="order_id",
    how="inner"
)

print(merged_df.shape)
merged_df.head()
```

Verify the Merged Dataset

```python
merged_df.info()
```

```python
merged_df.isnull().sum()
```

```python
merged_df.duplicated().sum()
```

Save the Merged Dataset

```python
merged_df.to_csv(
    "../datasets/cleaned/merged_customer_data.csv",
    index=False
)

print("Merged dataset saved successfully!")
```

## Dataset Merging

The cleaned datasets were merged using the following keys:

- Customers ↔ Orders using `customer_id`
- Orders ↔ Order Items using `order_id`
- Orders ↔ Payments using `order_id`

The resulting merged dataset contains customer details, order information, product prices, freight costs, and payment details in a single table. This consolidated dataset will be used for feature engineering and customer segmentation.
